```python
import scipy.stats as stats

# Número de observaciones (n = 15) -> Grados de libertad (df = n - 1)
df = 14

# Valor de t buscado
t_val = -0.75

# Calcular P(T <= -0.75) usando la función de distribución acumulada (CDF)
probabilidad = stats.t.cdf(t_val, df)

print(f"Probabilidad: {probabilidad:.4f}")
```

```python
stats.chi2.ppf(0.90, df=9)  # Arroja 14.68365.
```
