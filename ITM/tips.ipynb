```python
import seaborn as sns
import pandas as pd

# 1. Cargar el dataset
df = sns.load_dataset('tips')

# 2. Eliminar valores nulos para garantizar datos completos
df_limpio = df.dropna()

# 3. Tomar muestra aleatoria de exactamente 83 datos
# random_state asegura que la muestra sea reproducible
muestra = df_limpio.sample(n=83, random_state=42).reset_index(drop=True)

# Crear variable discreta: 1 si es fin de semana (Sat, Sun), 0 si es entre semana
muestra['fin_de_semana'] = muestra['day'].apply(lambda x: 1 if x in ['Sat', 'Sun'] else 0)

# 4. Guardar en un archivo .dat (separado por tabulaciones, estándar para .dat)
muestra.to_csv('tips_muestra_83.dat', sep='\t', index=False)

print(f"Datos guardados en 'tips_muestra_83.dat'. Dimensiones: {muestra.shape}")
```

| Variable | Descripción en el contexto | Tipo de variable |
| :--- | :--- | :--- |
| `total_bill` | Monto total de la cuenta del restaurante (en dólares) antes de propina. | Cuantitativa Continua |
| `tip` | Monto de la propina dejada por el cliente (en dólares). | Cuantitativa Continua |
| `size` | Número total de personas (comensales) en la mesa. | Cuantitativa Discreta |
| `fin_de_semana` | Variable indicadora binaria: 1 si la visita fue en fin de semana (Sábado/Domingo), 0 si fue entre semana. | Cuantitativa Discreta |
| `sex` | Género de la persona que realiza el pago de la cuenta (Male = Masculino, Female = Femenino). | Cualitativa Nominal |
| `smoker` | Indica si el cliente fuma o no en el restaurante (Yes = Sí, No = No). | Cualitativa Nominal |
| `day` | Día de la semana en que se consumió el servicio (Thur, Fri, Sat, Sun). Tiene un orden cronológico natural. | Cualitativa Ordinal |
| `time` | Momento del día en que se consumió el servicio (Lunch = Almuerzo, Dinner = Cena). Tiene un orden temporal natural. | Cualitativa Ordinal |

**Resumen de cumplimiento de requisitos:**
*   **4 Cuantitativas:** 2 Continuas (`total_bill`, `tip`) y 2 Discretas (`size`, `fin_de_semana`).
*   **4 Cualitativas:** 2 Nominales (`sex`, `smoker`) y 2 Ordinales (`day`, `time`).

```python
# Diccionario con los nuevos nombres
nuevos_nombres = {
    'total_bill': 'cuenta',
    'tip': 'propina',
    'sex': 'sexo',
    'smoker': 'fumador',
    'day': 'dia',
    'time': 'turno',
    'size': 'personas',
    'fin_de_semana': 'finde'
}

# Renombrar las columnas
muestra.rename(columns=nuevos_nombres, inplace=True)

# Guardar nuevamente el archivo .dat con los nuevos nombres
muestra.to_csv('tips_muestra_83.dat', sep='\t', index=False)

# Verificar el cambio
print(muestra.head())
```
