```python
import numpy as np
from scipy.stats import skew, kurtosis, norm, expon, uniform

np.random.seed(42)

# 1. Distribución Normal (Referencia)
data_norm = norm.rvs(size=10000)
print("Normal:")
print(f"  Skewness: {skew(data_norm):.3f}")
print(f"  Exceso de Curtosis: {kurtosis(data_norm):.3f}")

# 2. Distribución Exponencial (Asimetría positiva, colas pesadas)
data_exp = expon.rvs(size=10000)
print("\nExponencial:")
print(f"  Skewness: {skew(data_exp):.3f}")
print(f"  Exceso de Curtosis: {kurtosis(data_exp):.3f}")

# 3. Distribución Uniforme (Simétrica, colas ligeras)
data_unif = uniform.rvs(size=10000)
print("\nUniforme:")
print(f"  Skewness: {skew(data_unif):.3f}")
print(f"  Exceso de Curtosis: {kurtosis(data_unif):.3f}")
```
